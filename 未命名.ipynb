```python
import tushare as ts
import pandas as pd
import datetime, time
import sqlite3
import sqlite_data
```

```python
pro = ts.pro_api()
```

```python
#获取申万一级行业列表
df1 = pro.index_classify(level='L1', src='SW2021', fields='index_code, industry_name, level, industry_code, is_pub, parent_code, src')
#获取申万二级行业列表
df2 = pro.index_classify(level='L2', src='SW2021', fields='index_code, industry_name, level, industry_code, is_pub, parent_code, src')
#获取申万三级级行业列表
df3 = pro.index_classify(level='L3', src='SW2021', fields='index_code, industry_name, level, industry_code, is_pub, parent_code, src')
```

```python
df = pd.concat([df1,df2,df3], axis=0)
```
