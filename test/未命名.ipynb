```python
import tushare as ts
import pandas as pd
import datetime, time
import sqlite3
import sqlite_data
import time
```

```python
pro = ts.pro_api()
```

```python
from sqlite_data import DataApi
data_api = DataApi()
```

```python
pro.stock_mx(trade_date='20221109')
```

```python
trade_cal = pd.DataFrame(data_api.trade_cal(exchange='SSE', is_open='1', fields=['cal_date'])['cal_date'][::-1])
```

```python
trade_cal
```

```python
trade_cal['week_info'] = trade_cal.apply(lambda x: datetime.datetime.strptime(x['cal_date'], '%Y%m%d').date().isocalendar(), axis=1)
trade_cal['year'] = trade_cal.apply(lambda x: x['week_info'][0], axis=1)
trade_cal['week'] = trade_cal.apply(lambda x: x['week_info'][1], axis=1)
trade_cal['day'] = trade_cal.apply(lambda x: x['week_info'][2], axis=1)
```

```python
df = trade_cal.groupby(['year','week']).apply(lambda x: x[x['day']==x['day'].max()])
```
