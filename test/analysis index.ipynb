```python
import tushare as ts
import pandas as pd
import matplotlib.pyplot as plt
```

```python
pro = ts.pro_api()
df = pro.index_daily(ts_code='000001.SH')
df.head(1)
```

```python
df['trade_date'] = pd.to_datetime(df['trade_date'])
df2 = df.set_index('trade_date')
df2.head(1)
```

```python
df2.plot(y=['total_mv', 'pe'], secondary_y='pe')
plt.show()
```

```python
df2.plot(y=['total_mv', 'pb'], secondary_y='pb')
plt.show()
```
